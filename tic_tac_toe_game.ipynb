```python
def print_board(board):
    """Prints the current state of the game board."""
    print("\n")
    print(f" {board[0]} | {board[1]} | {board[2]} ")
    print("---|---|---")
    print(f" {board[3]} | {board[4]} | {board[5]} ")
    print("---|---|---")
    print(f" {board[6]} | {board[7]} | {board[8]} ")
    print("\n")

def check_win(board, player):
    """Checks if the specified player has won the game."""
    win_conditions = [
        [0, 1, 2], [3, 4, 5], [6, 7, 8], # Horizontal rows
        [0, 3, 6], [1, 4, 7], [2, 5, 8], # Vertical columns
        [0, 4, 8], [2, 4, 6]             # Diagonals
    ]
    return any(all(board[pos] == player for pos in condition) for condition in win_conditions)

def check_tie(board):
    """Checks if the board is full, resulting in a tie."""
    return all(space in ["X", "O"] for space in board)

def play_game():
    """Main function to run the Tic-Tac-Toe game loop."""
    # Initialize the board with positions 1 through 9
    board = [str(i) for i in range(1, 10)]
    current_player = "X"

    print("Welcome to Tic-Tac-Toe!")
    print_board(board)

    while True:
        try:
            # Prompt user for their move
            move = int(input(f"Player {current_player}, choose a position (1-9): ")) - 1

            # Validate input boundaries
            if move < 0 or move > 8:
                print("Invalid choice! Please pick a number between 1 and 9.")
                continue

            # Check if the chosen spot is already occupied
            if board[move] in ["X", "O"]:
                print("That spot is already taken! Try another one.")
                continue

        except ValueError:
            print("Invalid input! Please enter a valid number.")
            continue

        # Place the player's marker on the board
        board[move] = current_player
        print_board(board)

        # Check for game over conditions
        if check_win(board, current_player):
            print(f"🎉 Congratulations! Player {current_player} wins!")
            break

        if check_tie(board):
            print("🤝 It's a tie game!")
            break

        # Alternate turns between X and O
        current_player = "O" if current_player == "X" else "X"

if __name__ == "__main__":
    play_game()
```
